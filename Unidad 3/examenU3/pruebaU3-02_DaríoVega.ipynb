## Ejercicio 2 · Criterio evaluado: **c)** · 2 puntos

La función `suma_multiplos(n, k)` debe devolver la **suma de los múltiplos de `k` entre 1 y `n`, ambos incluidos**. Por ejemplo, `suma_multiplos(10, 3)` = 3 + 6 + 9 = 18. Tiene **dos errores** que se detectan verificando el funcionamiento del bucle.

**2.1** Corrige la función.

````python
def suma_multiplos(n, k):
    """Suma los múltiplos de k desde 1 hasta n (ambos incluidos)."""
    total = 0
    for i in range(1, n + 1):
        if i % k == 0:
            total += i
    return total
````

**2.2** Escribe **al menos 3 pruebas propias** con `assert` que verifiquen la función, incluyendo **un caso límite** (por ejemplo, que `n` sea múltiplo de `k`, o `n = 0`).

````python
# TU CÓDIGO AQUÍ (mínimo 3 assert, uno de ellos un caso límite)
assert suma_multiplos(0, 5) == 0   # intervalo vacío
assert suma_multiplos(8, 3) == 9   # múltiplos 3 y 6
assert suma_multiplos(4, 10) == 0   # no hay múltiplos
````

## Ejercicio 3 · Criterios evaluados: **d)** y **e)** · 3 puntos

Elige en cada caso **el bucle más adecuado** a la necesidad (uno de ellos exige `for` y el otro `while`) y escribe el código.

**3.1 `for` con salida anticipada.** Escribe `primer_mayor(datos, limite)`: devuelve el **primer elemento mayor que `limite`**, o `None` si no hay ninguno. Debe **dejar de recorrer** los datos en cuanto lo encuentre.

**3.2 `while` con número de repeticiones desconocido.** Escribe `intentos_hasta_acertar(secreto, intentos)`: `intentos` es un iterable con los números que va probando un jugador. Devuelve **cuántos intentos hicieron falta** hasta acertar `secreto`, o `None` si nunca acierta. Debes usar un **bucle `while`** y dejar de consumir intentos al acertar. *Pista: `siguiente = next(it, None)` devuelve `None` cuando se acaba el iterador.*

````python
def primer_mayor(datos, limite):
    for dato in datos:
        if dato > limite:
            return dato
    return None

def intentos_hasta_acertar(secreto, intentos):
    it = iter(intentos)
    contador = 0
    siguiente = next(it, None)
    while siguiente is not None:
        contador += 1
        if siguiente == secreto:
            return contador
        siguiente = next(it, None)
    return None
````

````python
# Comprobación (NO modificar)
consumidos = []
def gen1():
    for x in [1, 5, 9, 20]:
        consumidos.append(x)
        yield x

assert primer_mayor(gen1(), 4) == 5
assert consumidos == [1, 5], "primer_mayor debe dejar de recorrer en cuanto encuentra el valor"
assert primer_mayor([1, 2], 10) is None
assert primer_mayor([], 0) is None

consumidos.clear()
def gen2():
    for x in [3, 8, 5, 8]:
        consumidos.append(x)
        yield x

assert intentos_hasta_acertar(5, gen2()) == 3
assert consumidos == [3, 8, 5], "No debes consumir más intentos tras acertar"
assert intentos_hasta_acertar(5, [1, 2]) is None
assert intentos_hasta_acertar(7, [7]) == 1
print("✔ Ejercicio 3 superado (el profesorado comprueba el uso de for y while)")
````

**4.2 (aplicar)** Escribe `pares_con_suma(lista, objetivo)` que devuelva la **lista de tuplas `(a, b)`** con todos los pares de elementos de la lista, tomados **en el orden en que aparecen** (el primero antes que el segundo), cuya suma sea `objetivo`. Usa **bucles anidados**.

Ejemplo: `pares_con_suma([1, 2, 3, 4, 5], 6)` → `[(1, 5), (2, 4)]`

````python
def pares_con_suma(lista, objetivo):
    pares = []
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            if lista[i] + lista[j] == objetivo:
                pares.append((lista[i], lista[j]))
    return pares
````
