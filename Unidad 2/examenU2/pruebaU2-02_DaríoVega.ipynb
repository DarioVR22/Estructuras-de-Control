## Ejercicio 2 · Criterios evaluados: **c)** y **d)** · 2 puntos

La función `precio_final` debería aplicar un **10 % de descuento a los socios** y un **5 % adicional** (acumulable) si se compran **10 o más unidades**, sea o no socio. Tiene **dos errores**: uno de **sangrado** y otro en el **control de la variable** `descuento`. Corrígela.

```python
def precio_final(precio, es_socio, unidades):
    descuento = 0
    if es_socio:
        descuento += 0.10
    if unidades >= 10:
        descuento += 0.05
    return round(precio * (1 - descuento), 2)
```

```python
# Comprobación (NO modificar)
assert precio_final(100, True, 10) == 85.0, "Socio con 10 unidades: 15 % de descuento"
assert precio_final(100, False, 10) == 95.0, "No socio con 10 unidades: 5 %"
assert precio_final(100, True, 1) == 90.0
assert precio_final(100, False, 1) == 100.0
print("✔ Ejercicio 2 superado")
```

## Ejercicio 3 · Criterios evaluados: **f)** e **i)** · 2 puntos

Escribe la función `tarifa_parking(minutos)` que calcule el precio del aparcamiento:

| Tiempo aparcado | Precio |
|---|---|
| Hasta 15 minutos (incluido) | 0 € |
| Más de 15 y hasta 60 minutos (incluido) | 1.50 € |
| Más de 60 y hasta 180 minutos (incluido) | 4.00 € |
| Más de 180 minutos | 10.00 € |
| Valor negativo | `None` (dato no válido) |

```python
def tarifa_parking(minutos):
    if minutos < 0:
        return None
    elif 0 <= minutos <= 15:
        return 0
    elif 15 < minutos <= 60:
        return 1.5
    elif 60 < minutos <= 180:
        return 4.0
    else:
        return 10.0
```

```python
# Comprobación (NO modificar)
casos = {-5: None, 0: 0, 15: 0, 16: 1.5, 60: 1.5, 61: 4.0, 180: 4.0, 181: 10.0, 500: 10.0}
for minutos, esperado in casos.items():
    assert tarifa_parking(minutos) == esperado, f"Falla con {minutos} min: esperado {esperado}"
print("✔ Ejercicio 3 superado")
```

**4.2 (criterio j)** Escribe `acceso(edad, tiene_entrada, es_vip)` **usando condicionales anidadas** (un `if` dentro de otro) que devuelva:

- Si **no** tiene entrada → `"Sin entrada"`.
- Si tiene entrada y es **mayor de edad** (`edad >= 18`): `"Zona VIP"` si es VIP, o `"Zona general"` si no.
- Si tiene entrada y es **menor de edad**: `"Zona infantil"`.

```python
def acceso(edad, tiene_entrada, es_vip):
    if tiene_entrada:
        if edad >= 18:
            if es_vip:
                return "Zona VIP"
            else:
                return "Zona general"
        else:
            return "Zona infantil"
    else:
        return "Sin entrada"
```

## Ejercicio 5 · Criterio evaluado: **h)** · 1 punto

Reescribe las dos funciones en **forma compacta**, con una **única expresión condicional** (operador ternario) y sin usar `max()`:

- `tipo_numero(n)` → `"par"` o `"impar"`.
- `mayor(a, b)` → el mayor de los dos números.

```python
def tipo_numero(n):
    return "par" if n % 2 == 0 else "impar"

def mayor(a, b):
    return a if a > b else b
```

```python
# Comprobación (NO modificar)
assert tipo_numero(4) == "par" and tipo_numero(7) == "impar" and tipo_numero(0) == "par"
assert mayor(3, 9) == 9 and mayor(9, 3) == 9 and mayor(-2, -2) == -2
print("✔ Ejercicio 5 superado (el profesorado comprueba el uso del ternario)")
```
