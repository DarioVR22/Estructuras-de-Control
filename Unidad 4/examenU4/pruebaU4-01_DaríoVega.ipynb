**2.1 (criterio c)** Define la función `precio_con_descuento(precio, porcentaje=0, iva=0.21)` que devuelva `precio * (1 - porcentaje/100) * (1 + iva)` **redondeado a 2 decimales**. Debe incluir un **docstring** que explique qué hace y qué significan los parámetros.

````python
# TU CÓDIGO AQUÍ
def precio_con_descuento(precio, porcentaje=0, iva=0.21):
    """
    Parámetros: 
    precio: precio del articulo/producto
    pocentaje: porcentaje de descuento que recibe el producto, el valor por defecto es 0.
    iva: porcentaje de iva que va a recibir el producto, el valor por defecto es 0.21.
    
    Funcionamiento:
    Una vez dada a la función el precio de un producto esta le calcula otro precio añadiendole un descuento(0=0% por defecto) y un porcentaje de iva(0.21=21% por defecto). 
    """
    
    return round(precio * (1 - porcentaje/100) * (1 + iva),2)
````

**2.2 (criterio d)** Usando **funciones `lambda`** (con `sorted`, `filter` y `map`) obtén:

## Ejercicio 3 · Criterio evaluado: **e)** · 2 puntos

Disponemos de esta función ya definida (versión **iterativa**):

````python
def factorial_iter(n):
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado
````

**3.1** A partir de ella, escribe `factorial(n)` en versión **recursiva** (con su caso base y su caso recursivo).

**3.2** Escribe `suma_factoriales(n)`, también **recursiva**, que devuelva `0! + 1! + 2! + ... + n!` **reutilizando tu función `factorial`**. Ejemplo: `suma_factoriales(3)` = 1 + 1 + 2 + 6 = 10.

````python
def factorial(n):
    if n < 2:
        return 1
    return n * factorial(n - 1)

def suma_factoriales(n):
    if n == 0:
        return factorial(0)
    return factorial(n) + suma_factoriales(n - 1)
````

````python
# Comprobación (NO modificar)
for k in range(0, 10):
    assert factorial(k) == factorial_iter(k), f"factorial({k}) no coincide con la versión iterativa"
assert suma_factoriales(0) == 1
assert suma_factoriales(3) == 10
assert suma_factoriales(5) == 154
print("✔ Ejercicio 3 superado (el profesorado comprueba que es recursiva y reutiliza factorial)")
````
