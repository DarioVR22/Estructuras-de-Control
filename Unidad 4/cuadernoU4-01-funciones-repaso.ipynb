## 1. Necesidad y ventajas de las funciones (criterio a)

Una **función** es un bloque de código con nombre que realiza una tarea concreta y que se puede **reutilizar** llamándolo cuantas veces haga falta.

### El problema: código repetido

````python
# Sin funciones: el mismo cálculo copiado varias veces
precio1, precio2, precio3 = 100, 250, 80

total1 = precio1 + precio1 * 0.21
total2 = precio2 + precio2 * 0.21
total3 = precio3 + precio3 * 0.21
print(total1, total2, total3)

# Si el IVA cambia, hay que modificar CADA línea (y es fácil olvidar alguna)
````

````python
# Con una función: se escribe una vez y se reutiliza
def con_iva(precio, iva=0.21):
    return precio + precio * iva

print(con_iva(100), con_iva(250), con_iva(80))
print(con_iva(100, 0.10))      # IVA reducido
````

### Ventajas

| Ventaja | Explicación |
|---|---|
| **Reutilización** | Escribir una vez, usar muchas (principio **DRY**: *Don't Repeat Yourself*) |
| **Legibilidad** | Un nombre descriptivo explica *qué* hace el código |
| **Mantenimiento** | Un cambio en un solo lugar |
| **Modularidad** | Se divide un problema grande en partes pequeñas |
| **Pruebas** | Cada función se puede probar por separado |
| **Eficiencia** | Menos código, menos errores, equipos que trabajan en paralelo |

### Tipos de funciones que veremos

1. **Built-in** (incorporadas): ya vienen con Python.
2. **Definidas por el usuario** (`def`).
3. **Lambda** (anónimas, de una sola expresión).
4. **Recursivas** (se llaman a sí mismas).

## 3. Funciones definidas por el usuario (criterio c)

### 3.1 Sintaxis

```python
def nombre_funcion(parametro1, parametro2=valor_defecto):
    """Docstring: qué hace la función."""
    # cuerpo (sangrado)
    return resultado
```

- **Parámetros**: variables en la definición. **Argumentos**: valores en la llamada.
- `return` devuelve un valor y **termina** la función. Sin `return` se devuelve `None`.
- Nombres en `snake_case`, descriptivos y con verbo (`calcular_media`).

````python
def saludar(nombre):
    """Muestra un saludo (no devuelve nada)."""
    print(f"Hola, {nombre}")

def cuadrado(x):
    """Devuelve x elevado al cuadrado."""
    return x ** 2

saludar("Ana")
resultado = cuadrado(7)
print(resultado)

# Función sin return -> None
valor = saludar("Luis")
print(valor)
````

### 3.2 Tipos de argumentos

````python
def presentar(nombre, edad, ciudad="Madrid"):
    return f"{nombre}, {edad} años, de {ciudad}"

print(presentar("Ana", 30))                       # posicionales + valor por defecto
print(presentar("Luis", 25, "Sevilla"))
print(presentar(edad=40, nombre="Marta"))         # argumentos con nombre (keyword)
````

````python
# *args: número variable de argumentos posicionales
def sumar(*numeros):
    total = 0
    for n in numeros:
        total += n
    return total

print(sumar(1, 2), sumar(1, 2, 3, 4, 5), sumar())

# **kwargs: número variable de argumentos con nombre
def mostrar_ficha(**datos):
    for clave, valor in datos.items():
        print(f"  {clave}: {valor}")

mostrar_ficha(nombre="Ana", edad=30, ciudad="Madrid")
````

````python
# Devolver varios valores (en realidad, una tupla)
def estadisticas(lista):
    return min(lista), max(lista), sum(lista) / len(lista)

minimo, maximo, media = estadisticas([4, 8, 15, 16, 23, 42])
print(minimo, maximo, media)
````

````python
# Trampa clásica: valor por defecto mutable
def agregar_mal(elemento, lista=[]):
    lista.append(elemento)
    return lista

print(agregar_mal(1))
print(agregar_mal(2))      # [1, 2] -> ¡la lista se comparte entre llamadas!

def agregar_bien(elemento, lista=None):
    if lista is None:
        lista = []
    lista.append(elemento)
    return lista

print(agregar_bien(1))
print(agregar_bien(2))     # [2]
````

### 3.4 Funciones que utilizan estructuras de control

Las funciones son el sitio natural donde usar condicionales y bucles.

````python
def es_primo(n):
    """Devuelve True si n es primo."""
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    return True

def primos_hasta(limite):
    return [n for n in range(2, limite + 1) if es_primo(n)]

print(primos_hasta(40))
````

````python
def clasificar_nota(nota):
    """Convierte una nota numérica en calificación."""
    if not 0 <= nota <= 10:
        return "Nota no válida"
    if nota < 5:
        return "Suspenso"
    elif nota < 7:
        return "Aprobado"
    elif nota < 9:
        return "Notable"
    return "Sobresaliente"

for n in (3, 5.5, 7.5, 10, 12):
    print(n, "->", clasificar_nota(n))
````

````python
# Funciones que llaman a otras funciones (reutilización)
def media(lista):
    return sum(lista) / len(lista)

def desviacion(lista):
    m = media(lista)
    return (sum((x - m) ** 2 for x in lista) / len(lista)) ** 0.5

datos = [2, 4, 4, 4, 5, 5, 7, 9]
print(f"Media={media(datos)}  Desviación={desviacion(datos)}")
````

````python
# Funciones como argumentos (funciones de orden superior)
def aplicar(funcion, lista):
    return [funcion(x) for x in lista]

print(aplicar(abs, [-3, 4, -5]))
print(aplicar(str.upper, ["a", "b"]))
print(aplicar(es_primo, [2, 4, 7, 9]))
````

1. `es_par(n)` → `True` si `n` es par.
2. `contar_vocales(texto)` → número de vocales.
3. `descuento(precio, porcentaje=10)` → precio tras aplicar el descuento.
4. `es_palindromo(texto)` → `True` si el texto se lee igual al revés (ignora mayúsculas y espacios).
5. `maximo_y_posicion(lista)` → devuelve el máximo y su índice.

````python
# TU CÓDIGO AQUÍ (define las funciones)
def es_par(n):
    return n % 2 == 0

def contar_vocales(texto):
    contador = 0

    for letra in texto.lower():
        if letra in "aeiou":
            contador += 1

    return contador

def descuento(precio, porcentaje=10):
    precio = precio - (precio*porcentaje/100)
    return precio

def es_palindromo(texto):
    texto = texto.lower().replace(" ", "")
    return texto[::-1] == texto

def maximo_y_posicion(lista):
    maximo = max(lista)
    posicion = lista.index(maximo)

    return maximo, posicion

# Pruebas (descomenta cuando las tengas definidas)
assert es_par(4) and not es_par(7)
assert contar_vocales("Python es genial") == 5
assert descuento(200) == 180
assert es_palindromo("Anita lava la tina")
assert maximo_y_posicion([3, 9, 2]) == (9, 1)
print("¡Todo correcto!")
````

## 4. Funciones `lambda` (criterio d)

````python
cuadrado = lambda x: x ** 2
suma = lambda a, b: a + b
es_par = lambda n: n % 2 == 0
saludo = lambda nombre="mundo": f"Hola, {nombre}"

print(cuadrado(5), suma(2, 3), es_par(8), saludo(), saludo("Ana"))

# Equivalente con def:
def cuadrado_def(x):
    return x ** 2
````

### 4.1 Dónde tienen sentido las lambdas

Cuando una función pequeña se pasa **como argumento**: `sorted(key=)`, `map`, `filter`, `max/min(key=)`, `sort`.

````python
alumnos = [("Ana", 8.5), ("Luis", 6.0), ("Marta", 9.2), ("Pedro", 4.8)]

print(sorted(alumnos, key=lambda a: a[1]))                  # por nota
print(sorted(alumnos, key=lambda a: a[1], reverse=True))    # de mayor a menor
print(max(alumnos, key=lambda a: a[1]))                     # mejor alumno
print(min(alumnos, key=lambda a: len(a[0])))                # nombre más corto
````

````python
numeros = [1, 2, 3, 4, 5, 6, 7, 8]

print(list(map(lambda n: n * 10, numeros)))
print(list(filter(lambda n: n > 4, numeros)))

# Condicional dentro de una lambda (ternario)
etiqueta = lambda n: "par" if n % 2 == 0 else "impar"
print([etiqueta(n) for n in numeros[:4]])
````

````python
# Lambda que devuelve lambdas (cierre)
multiplicador = lambda n: (lambda x: x * n)
doble = multiplicador(2)
triple = multiplicador(3)
print(doble(7), triple(7))

# Diccionario de operaciones
operaciones = {
    "+": lambda a, b: a + b,
    "-": lambda a, b: a - b,
    "*": lambda a, b: a * b,
    "/": lambda a, b: a / b if b != 0 else None,
}
for simbolo, funcion in operaciones.items():
    print(f"8 {simbolo} 2 = {funcion(8, 2)}")
````

## 5. Funciones recursivas (criterio e)

Una función **recursiva** es la que **se llama a sí misma**. Necesita:

1. **Caso base**: situación en la que **no** hay más llamadas (evita la recursión infinita).
2. **Caso recursivo**: reduce el problema y se llama a sí misma con un problema "más pequeño".

```mermaid
flowchart TD
    A["factorial(4)"] --> B["4 * factorial(3)"]
    B --> C["3 * factorial(2)"]
    C --> D["2 * factorial(1)"]
    D --> E["factorial(1) = 1  (caso base)"]
```

````python
def factorial(n):
    if n <= 1:                    # caso base
        return 1
    return n * factorial(n - 1)   # caso recursivo

print(factorial(5))     # 120
````

````python
# Traza de la recursión: ver cómo se apilan las llamadas
def factorial_traza(n, nivel=0):
    sangria = "  " * nivel
    print(f"{sangria}factorial({n})")
    if n <= 1:
        print(f"{sangria}-> 1")
        return 1
    resultado = n * factorial_traza(n - 1, nivel + 1)
    print(f"{sangria}-> {resultado}")
    return resultado

factorial_traza(4)
````

### 5.1 Recursión partiendo de funciones definidas anteriormente

Una forma natural de crear una función recursiva es **partir de una versión iterativa o de una función ya escrita** y reformularla: *"¿cómo resuelvo el problema para `n` suponiendo que ya sé resolverlo para `n-1`?"*

````python
# Función definida anteriormente (iterativa)
def suma_hasta_iter(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

# Versión recursiva: suma_hasta(n) = n + suma_hasta(n-1)
def suma_hasta(n):
    if n <= 0:
        return 0
    return n + suma_hasta(n - 1)

assert suma_hasta(10) == suma_hasta_iter(10) == 55
print(suma_hasta(100))
````

````python
# Reutilizando funciones ya definidas: potencia recursiva
def potencia(base, exponente):
    if exponente == 0:
        return 1
    return base * potencia(base, exponente - 1)

# Cuadrado y cubo se construyen reutilizando 'potencia'
cuadrado_r = lambda x: potencia(x, 2)
cubo_r = lambda x: potencia(x, 3)
print(potencia(2, 10), cuadrado_r(9), cubo_r(3))
````

````python
# Suma de los dígitos de un número
def suma_digitos(n):
    if n < 10:
        return n
    return n % 10 + suma_digitos(n // 10)

print(suma_digitos(12345))      # 15

# Inversa de una cadena
def invertir(texto):
    if len(texto) <= 1:
        return texto
    return invertir(texto[1:]) + texto[0]

print(invertir("recursividad"))

# Un palíndromo reutilizando 'invertir'
def es_palindromo(texto):
    limpio = texto.replace(" ", "").lower()
    return limpio == invertir(limpio)

print(es_palindromo("Anita lava la tina"))
````

````python
# Fibonacci: recursión doble (¡ineficiente!)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print([fib(i) for i in range(10)])
````

````python
# Máximo común divisor (algoritmo de Euclides)
def mcd(a, b):
    if b == 0:
        return a
    return mcd(b, a % b)

print(mcd(48, 18))     # 6

# Búsqueda binaria recursiva
def buscar(lista, objetivo, inicio=0, fin=None):
    if fin is None:
        fin = len(lista) - 1
    if inicio > fin:
        return -1
    medio = (inicio + fin) // 2
    if lista[medio] == objetivo:
        return medio
    if lista[medio] < objetivo:
        return buscar(lista, objetivo, medio + 1, fin)
    return buscar(lista, objetivo, inicio, medio - 1)

print(buscar([1, 3, 5, 7, 9, 11], 7))
````

### 5.2 Límites de la recursión

Python limita la profundidad (por defecto ~1000 llamadas). Sin caso base, o con un problema demasiado grande, se produce `RecursionError`.

````python
import sys
print("Límite de recursión:", sys.getrecursionlimit())

def sin_caso_base(n):
    return sin_caso_base(n + 1)

try:
    sin_caso_base(0)
except RecursionError as e:
    print("RecursionError:", e)
````

## 6. Eficiencia: comparando versiones (criterio general del RA)

Las funciones permiten **mejorar la eficiencia** del programa: menos código, reutilización y, con técnicas como la **memoización**, menos cálculo.

````python
import time
from functools import lru_cache

def fib_lento(n):
    return n if n < 2 else fib_lento(n - 1) + fib_lento(n - 2)

@lru_cache(maxsize=None)
def fib_rapido(n):
    return n if n < 2 else fib_rapido(n - 1) + fib_rapido(n - 2)

def fib_iterativo(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

for nombre, f in [("recursivo simple", fib_lento), ("recursivo + caché", fib_rapido), ("iterativo", fib_iterativo)]:
    inicio = time.perf_counter()
    resultado = f(30)
    duracion = time.perf_counter() - inicio
    print(f"{nombre:20} fib(30) = {resultado}  ({duracion:.6f} s)")
````

**Conclusión:** la recursión es elegante para problemas con estructura recursiva (árboles, divide y vencerás), pero puede ser menos eficiente que un bucle. Se puede mejorar con memoización o convirtiéndola en iterativa.

### ✏️ Actividad 5
1. Escribe `contar_atras(n)` recursiva que imprima de `n` hasta 0.
2. Escribe `suma_lista(lista)` recursiva (suma del primer elemento + suma del resto).
3. Parte de esta función iterativa y escribe su versión recursiva `contar_vocales_rec(texto)`.
4. Escribe `binario(n)` recursiva que devuelva la representación binaria de `n` como cadena.

````python
# Función de partida (iterativa)
def contar_vocales_iter(texto):
    total = 0
    for c in texto.lower():
        if c in "aeiou":
            total += 1
    return total


# 1) Contar hacia atrás
def contar_atras(n):
    if n < 0:
        return

    print(n)
    contar_atras(n - 1)


# 2) Sumar los elementos de una lista
def suma_lista(lista):
    if len(lista) == 0:
        return 0

    return lista[0] + suma_lista(lista[1:])


# 3) Contar vocales recursivamente
def contar_vocales_rec(texto):
    texto = texto.lower()

    if len(texto) == 0:
        return 0

    if texto[0] in "aeiou":
        return 1 + contar_vocales_rec(texto[1:])

    return contar_vocales_rec(texto[1:])


# 4) Convertir decimal a binario
def binario(n):
    if n < 2:
        return str(n)

    return binario(n // 2) + str(n % 2)
````

````python

# ==========================================
# MINI-LIBRERÍA DE UTILIDADES
# ==========================================

# 1. Función propia con parámetro por defecto y docstring
def aplicar_iva(precios, iva=0.21):
    """Devuelve una lista de precios con el IVA aplicado."""
    return [round(p * (1 + iva), 2) for p in precios]


# 2. Función que utiliza filter y lambda
def baratos(precios, limite):
    """Devuelve los precios menores o iguales al límite."""
    return list(filter(lambda p: p <= limite, precios))


# 3. Función recursiva
def cuenta_regresiva(n):
    """Devuelve una lista desde n hasta 0 de forma recursiva."""
    return [0] if n == 0 else [n] + cuenta_regresiva(n - 1)


# 4. Función adicional: utilizar map y lambda
def duplicar_numeros(numeros):
    """Devuelve una lista con todos los números duplicados."""
    return list(map(lambda n: n * 2, numeros))


# 5. Función adicional: calcular la media
def calcular_media(numeros):
    """Devuelve la media de una lista de números."""
    return sum(numeros) / len(numeros)


# ==========================================
# BLOQUE DE PRUEBAS
# ==========================================

assert aplicar_iva([100]) == [121.0]
assert aplicar_iva([100], 0.10) == [110.0]

assert baratos([5, 20, 8], 10) == [5, 8]

assert cuenta_regresiva(3) == [3, 2, 1, 0]

assert duplicar_numeros([2, 4, 6]) == [4, 8, 12]

assert calcular_media([10, 20, 30]) == 20

print("Librería verificada ✔")
````
